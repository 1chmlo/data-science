```python
import pandas as pd
import os

# 1. Crear una carpeta para mantener ordenados los archivos exportados
os.makedirs('datasets_hospitales', exist_ok=True)

# 2. Definir el código del hospital que buscas
codigo_hospital = '200122'

# REVISIÓN IMPORTANTE: Reemplaza 'IdEstablecimiento' por el nombre exacto 
# que tenga la columna de los códigos de hospitales en tu dataset original df.
# (A veces se llama 'CodigoHospital', 'Establecimiento' o 'IdCentro')
columna_hospital = 'EstablecimientoCodigo'  # Cambia esto según tu dataset

# 3. Filtrar el dataframe
# Transformamos la columna a texto (str) por seguridad, para evitar 
# que no coincidan si uno es número (int) y otro es texto (object).
df_hospital = df[df[columna_hospital].astype(str) == codigo_hospital].copy()

# 4. Comprobar si encontró datos y exportar
cantidad_registros = len(df_hospital)

if cantidad_registros > 0:
    print(f"¡Éxito! Se encontraron {cantidad_registros} registros para el hospital {codigo_hospital}.")
    
    # Ordenar cronológicamente (opcional, pero recomendado para forecasting)
    if 'Anio' in df_hospital.columns and 'SemanaEstadistica' in df_hospital.columns:
        df_hospital = df_hospital.sort_values(by=['Anio', 'SemanaEstadistica'])
    
    # Crear el nombre del archivo usando f-strings
    nombre_archivo = f"datasets_hospitales/dataset_hospital_{codigo_hospital}.parquet"
    
    # Exportar el resultado a parquet sin el índice
    df_hospital.to_parquet(nombre_archivo, index=False)
    print(f"Dataset guardado en: {nombre_archivo}")
    
else:
    print(f"Advertencia: No se encontraron registros para el hospital con código {codigo_hospital}. Revisa el nombre de la columna o si el código existe en los datos.")
```
